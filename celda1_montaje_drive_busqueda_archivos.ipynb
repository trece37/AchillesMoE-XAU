```python
import os
from google.colab import drive

# 1. MONTAR GOOGLE DRIVE
# Esto te pedirá permiso y abrirá una ventana para autenticar tu cuenta.
drive.mount('/content/drive')
print("\n--- Google Drive montado exitosamente. ---")

# 2. RUTA DE TU PROYECTO
# ⚠️ IMPORTANTE: Reemplaza 'MiCarpetaProyecto' por el nombre real de tu proyecto
# La ruta base es siempre: '/content/drive/MyDrive/'
FOLDER_PATH = '/content/drive/MyDrive/AchillesTraining'

# 3. FUNCIÓN PARA IMPRIMIR LA ESTRUCTURA EN FORMA DE ÁRBOL
def print_directory_tree(startpath):
    """Recorre y muestra la estructura de carpetas y archivos."""
    print(f"\n--- Estructura de la carpeta '{os.path.basename(startpath)}' ---")

    if not os.path.exists(startpath):
        print(f"❌ ERROR: La ruta no existe: {startpath}")
        print("Asegúrate de haber escrito correctamente el nombre de la carpeta.")
        return

    for root, dirs, files in os.walk(startpath):
        # Calcula el nivel de profundidad de la carpeta actual
        # y elimina la ruta de inicio para calcular la indentación
        relative_path = root.replace(startpath, '', 1)
        level = relative_path.count(os.sep)

        # Define la indentación: 4 espacios por nivel de profundidad
        indent = ' ' * 4 * level

        # Imprime el nombre de la carpeta actual
        print(f'{indent}├── {os.path.basename(root)}/')

        # Imprime los archivos dentro de la carpeta actual
        subindent = ' ' * 4 * (level + 1)
        for f in files:
            print(f'{subindent}├── {f}')

# EJECUCIÓN
print_directory_tree(FOLDER_PATH)
```
